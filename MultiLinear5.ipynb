```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import seaborn as sb

df=pd.DataFrame({
    "ec":[1000,1150,1500,1700,2000,1900,1500,1400],
    "m":[25,24,22,20,21,21,18,20],
    "sc":[4,6,7,4,7,7,7,4],
    "ng":[5,5,5,5,5,5,5,5],
    "ftc":[50,45,55,47,50,45,49,50],
    "price":[800000,1000000,1400000,1300000,2000000,3500000,6000000,5000000]
})

x=df.iloc[:,:-1]
y=df.price
lg=LinearRegression()
lg.fit(x,y)
print(lg.score(x,y))
print(lg.coef_)
sb.pairplot(df)
```
