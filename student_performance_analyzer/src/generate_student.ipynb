```python
import pandas as pd
import numpy as np
import random

np.random.seed(42)

# Number of records
n = 300

# Sample data
names = ["Aarav", "Vivaan", "Aditya", "Vihaan", "Arjun", "Sai", "Reyansh",
         "Krishna", "Ishaan", "Atharv", "Anaya", "Diya", "Ira", "Myra",
         "Aadhya", "Saanvi", "Kavya", "Pihu"]

cities = ["Delhi", "delhi", "DELHI", "Mumbai", "mumbai", "Ahmedabad", "ahmedabad"]

data = {
    "StudentId": range(1, n + 1),
    "Name": np.random.choice(names, n),
    "Age": np.random.randint(18, 26, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "City": np.random.choice(cities, n),
    "StudyHours": np.round(np.random.normal(5, 2, n), 1),
    "Attendance": np.random.randint(60, 101, n)
}

df = pd.DataFrame(data)

# Generate marks logically
df["FinalMarks"] = (
    df["StudyHours"] * 8 +
    df["Attendance"] * 0.4 +
    np.random.normal(0, 10, n)
)

# Clip marks between 0 and 100
df["FinalMarks"] = df["FinalMarks"].clip(0, 100).round(2)

# Introduce missing values intentionally
for col in ["StudyHours", "Attendance"]:
    df.loc[df.sample(frac=0.05).index, col] = np.nan

# Save to CSV
df.to_csv("students.csv", index=False)

print("students.csv generated successfully!")
```
